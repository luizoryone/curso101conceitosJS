# 1️⃣ **Conceito Geral de Programação**

Programação é o processo de **dar instruções para um computador executar tarefas**.
É como escrever uma regra clara e estruturada que o computador possa seguir sem ambiguidade.

Você programa para que uma máquina:

* receba informações,
* processe essas informações,
* tome decisões,
* execute ações,
* e produza resultados previsíveis.

Um computador só entende instruções extremamente precisas.
Por isso, programar é **organizar o pensamento** e expressá-lo de forma lógica.

---

# 2️⃣ **Histórico e Evolução do Conceito**

### ✔ Primeira fase

Programação era feita com **cartões perfurados** e instruções extremamente básicas.

### ✔ Segunda fase

### ✔ Terceira fase

A programação passou a ser usada massivamente em sistemas corporativos, bancos, governo, indústria e internet.

### ✔ Hoje

Programação é considerada uma **linguagem universal de resolução de problemas**.

# 3️⃣ **Semântica e Sintaxe da Programação (em visão geral)**

* **Sintaxe**
  É a forma correta de escrever um comando.
  Ex.: Parênteses, chaves, pontos, vírgulas, ordem, etc.

* **Semântica**
  É o **significado** do comando, o que ele faz.

Exemplo simples:

```js
console.log("Olá");
```

**Sintaxe:**

* “console”, “.”, “log”, parênteses, aspas.

**Semântica:**

* imprimir uma mensagem no console.

# 2️⃣ **História do JavaScript — Linha do Tempo Didática**

### **1995 — Origem**

* Criada por *Brendan Eich* em apenas 10 dias.
* Nome inicial: Mocha, depois LiveScript.

### **Motivo de sua criação**

Os sites eram **estáticos**.
Navegador precisava de uma linguagem para interação.

### **1997 — Padronização**

* A linguagem passou a ser chamada oficialmente de **ECMAScript (ES)**.
* JavaScript é uma implementação do ECMAScript.

### **2009 — Node.js**

* Um marco histórico.
* Permite rodar JavaScript fora do navegador.
* Torna o JS uma linguagem **full stack**.

### **2015 — ES6**

A maior revolução da linguagem:

* `let`, `const`
* classes
* arrow functions
* template strings
* módulos
* promessas

JS moderno começa aqui.

### **Hoje**

* Considerada uma das linguagens mais versáteis do planeta.
* Core de quase toda tecnologia web.

# **📘 TÓPICO 4 — Variáveis, Escopo e Hoisting (JavaScript aplicado ao Sistema da Organização Pública Fiscalizatória)**

Este é um dos pilares centrais da linguagem.
Depois deste tópico, você entenderá completamente **como a linguagem guarda valores**, **onde eles existem** e **quando podem ser usados**.

---

# **1. Conceito Geral**

Variáveis são “caixas” onde armazenamos valores.

Em JavaScript, existem três formas principais de declarar variáveis:

| Palavra-chave | Pode mudar?          | Escopo | Hoisting                    | Uso recomendado           |
| ------------- | -------------------- | ------ | --------------------------- | ------------------------- |
| **var**       | Sim                  | Função | Sofre hoisting              | NÃO usar (herança antiga) |
| **let**       | Sim                  | Bloco  | Sofre hoisting (sem acesso) | Usada 95% das vezes       |
| **const**     | Não (valor imutável) | Bloco  | Sofre hoisting (sem acesso) | Valores fixos / objetos   |

---

# **2. Escopo**

Escopo é *onde* a variável existe.

### Tipos:

* **Escopo Global** → visível em todo o código
* **Escopo de Função** → visível apenas dentro da função
* **Escopo de Bloco** → visível dentro de `{ }`

Exemplo do sistema:

```js
let setor = "Tecnologia"; // escopo global

function cadastrarFuncionario() {
  let nome = "Carla"; // escopo da função
  if (true) {
    let cargo = "Auditoria"; // escopo de bloco
  }
}
```

---

# **3. Hoisting**

Hoisting é o comportamento do JavaScript de **elevar declarações ao topo do escopo**.

* `var` é elevada **com valor undefined**
* `let` e `const` são elevadas mas **não podem ser usadas** antes da linha de declaração
  → entram na “temporal dead zone”

Exemplo *problemático*:

```js
console.log(nome); // erro!
let nome = "Francisco";
```

---

# **4. Como isso se aplica ao tema da Organização Pública**

Cada variável no sistema representa um elemento do ambiente real:

* `const setores` → lista fixa dos setores existentes
* `let funcionarioAtivo` → status que muda
* `let equipamentoEmManutencao` → usado dentro de funções
* `const CONFIG_GERAL` → parâmetros do departamento de Tecnologia

O controle correto de escopo permite:

✔ segurança de dados
✔ menos bugs
✔ organização do sistema
✔ funções independentes
✔ reuso de lógica

---

# **5. Resenha**

Ao trabalhar com múltiplos setores da organização — tecnologia, auditoria, fiscalização, contratos, RH — você precisará lidar com dezenas de variáveis ao mesmo tempo.
Saber **onde elas existem** e **quando podem ser usadas** evita erros como:

* sobrescrever dados
* acessar informações antes do tempo
* alterar valores que deveriam ser constantes
* criar dependência entre funções

O escopo é a “fronteira lógica” que mantém o sistema organizado.

# **6. Q&A — Perguntas e Respostas**

### **1️⃣ Quando usar `let`?**

Sempre que o valor puder mudar.

### **2️⃣ Quando usar `const`?**

Sempre que você **não quiser permitir mudança**.
Ex.: lista fixa de setores, configurações gerais.

### **3️⃣ Objetos declarados com `const` podem ser alterados?**

Sim.
`const` protege a *referência*, não o conteúdo.

### **4️⃣ `var` deve ser usado hoje?**

Não.
`var` causa problemas de escopo e hoisting.

### **5️⃣ O que é a “temporal dead zone”?**

É o período em que uma variável `let`/`const` existe, mas **não pode ser usada** até ser declarada.

# **7. Abstração**

Variáveis podem ser vistas como gavetas dentro da organização:

* Há gavetas grandes (globais)
* Gavetas exclusivas de setores (função)
* Gavetas dentro de caixas específicas (blocos)

Hoisting seria algo como:
🗂 **"A gaveta é criada, mas só coloca algo nela na hora certa."**
